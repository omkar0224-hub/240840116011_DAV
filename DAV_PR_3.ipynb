**Aim: Apply filtering, sorting, aggregation, and GroupBy operations on real-world datasets.**

```python
# Import the Pandas library
# Pandas is used to work with data in table format
import pandas as pd
```

```python
# Create a dictionary to store employee data
# Each key becomes a column name
# Each list contains the values of that column
data = {
    "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name":['Rahul','Priya','Amit','Riya','Karan','Neha','Vivek','Pooja'],
    "Department":['IT','HR','IT','Finance','HR','Finance','IT','HR'],
    "City":['Surat','Navsari','Surat','Valsad','Surat','Navsari','Valsad','Surat'],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
# Create a DataFrame named df
# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)
```

**Step 1: Display the Dataset**

```python
# Display the dataset
df
```

**Step 2: Filter Data**

```python
# Display employees working in the IT department
df[df['Department']=='IT']
```

Display Employees with salary Greater than 50000

```python
# Display employees whose salary is greater than 50000
df[df['Salary']>50000]
```

Display Employees from Surat

```python
# Display employees who are from Surat
df[df['City']=='Surat']
```

Apply Multiple Conditions

```python
# Display IT employees whose salary is greater than 50000
df[(df['Department']=='IT') & (df['Salary']>50000)]
```

**Step 2: Sort Data**

Sort Salary in Ascending Order

```python
# Sort employees by salary from lowest to highest
df.sort_values(by='Salary')
```

Sort Salary in Descending Order

```python
# Sort employees by salary from highest to lowest
df.sort_values(by=['Salary'], ascending=[False])
```

Sort by Department and Salary

```python
# Sort first by Department and then by Salary
df.sort_values(by=['Department','Salary'], ascending=[True,False])
```

**Step 4: Perform Aggregation**

Calculate Total Salary

```python
# Calculate the total salary of all employees
df['Salary'].sum()
```

Calculate Average salary

```python
# Calculate the average salary
df['Salary'].mean()
```

Find highest Salary

```python
# Find the highest salary
df['Salary'].max()
```

Find lowest Salary

```python
# Find the lowest salary
df['Salary'].min()
```

Count Employees

```python
# Count the total number of employees
df['Emp_ID'].count()
```

**Step 5: Perform GroupBY Opertaions**



Average Salary Department-Wise

```python
# Calculate average salary for each department
df.groupby('Department')['Salary'].mean()
```

Minimum Salary Department-Wise

```python
# Find minimum salary in each department
df.groupby('Department')['Salary'].min()
```

Count Total Employees Department-Wise

```python
# Count employees in each department
df.groupby('Department')['Emp_ID'].count()
```

Total Salary Department-Wise

```python
# Calculate total salary for each department
df.groupby('Department')['Salary'].sum()
```

Maximum Salary Department-Wise

```python
# Find the highest salary in each department
df.groupby('Department')['Salary'].max()
```

Group by City

```python
# City Wise Average Salary
df.groupby('City')['Salary'].mean()
```

**Step 6: Multiple Aggregation Function**

```python
# Perform multiple calculations on salary department-wise
df.groupby('Department')['Salary'].agg(['count','sum','mean','min','max'])
```

**Step 7: Save the Result to CSV File and Download CSV File**

```python
# Save the DataFrame as a CSV file
df.to_csv('Employee_Data.csv', index=False)

# Import Google Colab file handling
from google.colab import files

# Download the CSV file
files.download('Employee_Data.csv')
```
