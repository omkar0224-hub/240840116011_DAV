**Aim: Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON files using Pandas.**

**Step 1: Import pandas**

```python
# Import the Pandas library
# Pandas is used to work with tables (DataFrames)
import pandas as pd
```

**Step 2:Create Sample Data**

```python
# Create a dictionary to store student data
# Each key (ID, Name, Age, etc.) becomes a column
# Each list becomes the values of that column
data = {
    "ID": [1,2,3,4,5,6],
    "Name": ["Rahul", "Priya", "Amit", "Karan", "Riya", "Aman"],
    "Age": [25, 30, 35, 40, 45, 50],
    "City": ["Surat", "Navsari", "Valsad", "Surat", "Gandhinagar", "Bharuch"],
    "Marks": [90, 85, 80, 75, 70, 65]
}
```

**Step 3: Create a Dataframe**

```python
# Creating a dataframe named df
# Convert the dictionary into a Pandas DataFrame (table)
df = pd.DataFrame(data)

# Display the DataFrame
df
```

**Step 4: Save Data As a CSV File**

```python
# Save the DataFrame as a CSV file

# index=False means do not save row numbers in the CSV file
df.to_csv("students.csv", index=False)

# Print a success message
print("CSV file created successfully")
```

**Step 5: Save Data as an Excel File**

```python
# Save the DataFrame as an Excel file

# index=False means do not save row numbers in the Excel file
df.to_excel("students.xlsx", index=False)

# Print a success message
print("Excel file created successfully")
```

**Step 6: Save Data as a JSON File**

```python
# Save the DataFrame as a JSON file

# orient="records" stores each row as a separate JSON object
# indent=4 adds spaces to make the JSON file easy to read
df.to_json("students.json", orient="records", indent=4)

# Print a success message
print("JSON file created successfully")
```

**Step 7: Read the CSV file**

```python
# Read the CSV file into a DataFrame
df_csv = pd.read_csv("students.csv")

# Display the DataFrame
df_csv
```
